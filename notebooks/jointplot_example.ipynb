```python
import pandas as pd 
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('data/tmp2.csv')
```

```python
df = df.replace('[^0-9]+', np.nan, regex=True)
df = df.astype(float)
```

```python
column_name1 = "PTA right 4000Hz AC"
column_name2 = "PTA right 2000Hz AC"
```

```python
df_without_nan = df.dropna(subset=[column_name1, column_name2])
```

## Вариант 1

Ребятам обязательно нужно на графиках:
- функция распределения плотности (сглаженная гистограмка)
- сами точки 
- подписи осей

Желательно добавить: 
- аппроксимировать кривой (линией)
- корреляция Пирсона или $R^2$ (коэффициент детерминации)

```python
r, _ = stats.pearsonr(df_without_nan[column_name1], df_without_nan[column_name2])
```

```python
g = sns.jointplot(data=df, x="PTA right 4000Hz AC", y="PTA right 2000Hz AC", kind='reg') #kde

g.ax_joint.annotate(f'$R = {r:.3f}$',
                    xy=(0.1, 0.9), xycoords='axes fraction',
                    ha='left', va='center',
                    bbox={'boxstyle': 'round', 'fc': 'powderblue', 'ec': 'navy'})

g.ax_joint.scatter(df[column_name1], df[column_name2])

# g.ax_joint.lmplot(data=df, x="PTA right 4000Hz AC", y="PTA right 2000Hz AC", height=3, aspect=1)
sns.regplot(data=df, x="PTA right 4000Hz AC", y="PTA right 2000Hz AC", ax=g.ax_joint)

plt.tight_layout()
```

## Вариант 2

`JointGrid` -- то же самое, что `jointplot`, но более гибкий, я не смогла убрать столбчатую гиастограмму из `jointplot`, но при этом оставить прямую, поэтому собрала 2 вариант из `JointGrid`

```python
x, y = df[column_name1], df[column_name2]
```

```python
g = sns.JointGrid()

x, y = df[column_name1], df[column_name2]

sns.scatterplot(x=x, y=y, s=100, linewidth=1.5, ax=g.ax_joint)
sns.regplot(x=x, y=y, ax=g.ax_joint)
g.ax_marg_x.set_xlim(0)
sns.kdeplot(y=y, linewidth=2, ax=g.ax_marg_y)
sns.kdeplot(x=x, linewidth=2, ax=g.ax_marg_x)

g.ax_joint.annotate(f'$R = {r:.3f}$',
                    xy=(0.1, 0.9), xycoords='axes fraction',
                    ha='left', va='center',
                    bbox={'boxstyle': 'round', 'fc': 'powderblue', 'ec': 'navy'})
```
